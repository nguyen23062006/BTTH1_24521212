# BÁO CÁO THUẬT TOÁN NEWTON–RAPHSON

## 1. Mục tiêu

Sử dụng thuật toán Newton–Raphson để tìm điểm dừng của hàm:

$$
f(x,y)=x^4+2xy+y^2
$$

Điểm khởi tạo được chọn là:

$$
u_0=(10,10)
$$

```python
import numpy as np
```

## 2. Cơ sở lý thuyết

Gradient của hàm mục tiêu:

$$
\nabla f(x,y)=
\begin{bmatrix}
4x^3+2y\\
2x+2y
\end{bmatrix}
$$

Ma trận Hessian:

$$
H(x,y)=
\begin{bmatrix}
12x^2 & 2\\
2 & 2
\end{bmatrix}
$$

Công thức cập nhật Newton:

$$
u_{k+1}=u_k-H(u_k)^{-1}\nabla f(u_k)
$$

Thuật toán dừng khi chuẩn của gradient nhỏ hơn một ngưỡng cho trước.

```python
def grad(u):
  x, y = u
  return np.array([
      4*x**3 + 2*y,
      2*x + 2*y
  ])
```

```python
def hessian(u):
  x, y = u
  return np.array([
    [12*x**2, 2],
    [2, 2]
  ])
```

## 3. Cài đặt thuật toán


```python
import copy
def newton_optimizer(grad_fn, hessian_fn, u_0, tol=1e-8, max_iter=1000):
  u = copy.deepcopy(np.array(u_0, dtype=float))
  history = [copy.deepcopy(u)]

  for _ in range(max_iter):
    grad = grad_fn(u)

    if np.linalg.norm(grad) <= tol:
        break

    hessian = hessian_fn(u)
    a = np.linalg.solve(hessian, grad)
    u = u - a
    history.append(copy.deepcopy(u))
  return history, u

```

## 4. Chạy thuật toán

```python
u_0 = (10, 10)
history, u = newton_optimizer(grad, hessian, u_0, max_iter=1000)
print("Điểm cực trị:", u)
```
